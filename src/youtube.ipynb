```python
def clean_transcript(raw_subtitle_content):
    """
    Cleans raw VTT/SRT subtitle content to return only the spoken text.
    """
    if not raw_subtitle_content:
        return ""
    lines = raw_subtitle_content.splitlines()
    cleaned_lines = []
    for line in lines:
        line = line.strip()
        if not line or "-->" in line or line.upper() == "WEBVTT" or re.fullmatch(r'\d+', line):
            continue
        line = re.sub(r'<[^>]+>', '', line)
        line = re.sub(r'\s+', ' ', line).strip()
        if line:
            cleaned_lines.append(line)
    return " ".join(cleaned_lines)

def download_subtitle(video_url, video_id):
    """
    Downloads subtitles for a given video and returns the raw content.
    """
    temp_dir = tempfile.mkdtemp()
    ydl_opts_subtitles = {
        'writesubtitles': True,
        'writeautomaticsub': True,
        'subtitleslangs': ['en'],
        'skip_download': True,
        'outtmpl': os.path.join(temp_dir, f'{video_id}.%(ext)s'),
        'quiet': True,
    }
    raw_subtitle_content = ""
    try:
        with yt_dlp.YoutubeDL(ydl_opts_subtitles) as ydl:
            ydl.download([video_url])
        for file in os.listdir(temp_dir):
            if file.startswith(video_id) and file.endswith(('.vtt', '.srt')):
                with open(os.path.join(temp_dir, file), 'r', encoding='utf-8') as f:
                    raw_subtitle_content = f.read()
                break
    except Exception as e:
        print(f"  -> Could not download subtitles: {e}")
    finally:
        shutil.rmtree(temp_dir)
    return raw_subtitle_content

def scrape_playlist_metadata(playlist_url):
    """
    Extracts comprehensive metadata and cleaned subtitles for all videos in a playlist.
    """
    # Step 1: Get a flat list of all video IDs in the playlist
    playlist_ydl_opts = {'quiet': True, 'extract_flat': True}
    with yt_dlp.YoutubeDL(playlist_ydl_opts) as ydl:
        playlist_info = ydl.extract_info(playlist_url, download=False)

    video_entries = playlist_info.get('entries', [])
    if not video_entries:
        print("Could not find any videos in the playlist.")
        return None

    print(f"\nFound {len(video_entries)} videos. Extracting metadata and subtitles...\n")

    all_video_data = []

    # Step 2: Loop through each video and extract its full metadata
    with yt_dlp.YoutubeDL({'quiet': True, 'skip_download': True}) as ydl:
        for i, entry in enumerate(video_entries):
            video_id = entry.get('id')
            if not video_id:
                continue
                
            video_url = f"https://www.youtube.com/watch?v={video_id}"
            print(f"Processing video {i+1}/{len(video_entries)}: {entry.get('title', video_id)}")

            try:
                # This call fetches the full metadata for the individual video
                video_info = ydl.extract_info(video_url, download=False)

                # Download and clean the subtitles
                raw_subtitles = download_subtitle(video_url, video_id)
                cleaned_subtitles = clean_transcript(raw_subtitles)

                # Collect all desired metadata fields
                all_video_data.append({
                    'video_id': video_info.get('id'),
                    'webpage_url': video_info.get('webpage_url'),
                    'title': video_info.get('title'),
                    'upload_date': video_info.get('upload_date'),
                    'duration_seconds': video_info.get('duration'),
                    'view_count': video_info.get('view_count'),
                    'like_count': video_info.get('like_count'),
                    'comment_count': video_info.get('comment_count'),
                    'channel': video_info.get('channel'),
                    'channel_url': video_info.get('channel_url'),
                    'uploader': video_info.get('uploader'),
                    'description': video_info.get('description'),
                    'tags': ', '.join(video_info.get('tags', [])),
                    'categories': ', '.join(video_info.get('categories', [])),
                    'thumbnail_url': video_info.get('thumbnail'),
                    'cleaned_subtitles': cleaned_subtitles
                })
                
                # Add a small delay to avoid rate limiting
                time.sleep(2)
                
            except Exception as e:
                print(f"  -> Failed to process video {video_id}. Reason: {e}")
                continue

    # Step 3: Save all collected data to a DataFrame
    df = pd.DataFrame(all_video_data)
    return df

print("Helper functions defined.")
```

```python
# Extract metadata and subtitles for the playlist
print("\nExtracting metadata and subtitles...")
print("="*60)

df = scrape_playlist_metadata(PLAYLIST_URL)

if df is not None and len(df) > 0:
    # Save to Excel
    output_filename = '../data/youtube_storyline_transcripts.xlsx'
    df.to_excel(output_filename, index=False)
    
    print(f"\n✅ Successfully scraped {len(df)} videos!")
    print(f"✅ Metadata and subtitles saved to {output_filename}")
    print(f"✅ Audio files saved to {AUDIO_DIR}/")
    print(f"\nColumns: {list(df.columns)}")
    print(f"\nFirst few titles:")
    for idx, title in enumerate(df['title'].head(5), 1):
        print(f"  {idx}. {title}")
else:
    print("\n❌ No data was scraped. Please check the playlist URL and try again.")
```
